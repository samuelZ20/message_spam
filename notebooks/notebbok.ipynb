```python
import matplotlib.pyplot as plt

# Fazer gráfico da quantidade de mensagens spam e não spam
plt.figure(figsize=(6, 4))
plt.bar(dados["label"].value_counts().index, dados["label"].value_counts().values)
plt.title("Quantidade de mensagens spam e não spam")
plt.xlabel("Label")
plt.ylabel("Quantidade")
plt.show()
```

LOGO, HÁ UMA QUANTIDADE DE MENSAGENS VERÍDICAS MAIOR QUE MENSAGENS SPAM

```python
#transformar em letras minúsculas
dados["sms_text"] = dados["sms_text"].str.lower()
#remover espaços em branco no início e no final
dados["sms_text"] = dados["sms_text"].str.strip()
#remover espaços desnecessários entre as palavras
dados["sms_text"] = dados["sms_text"].str.replace(r'\s+', ' ', regex=True)

dados.head(10)
```

AS PONTUAÇÕES, EMOJIS, CARACTERES podem influenciar se as mensagens são spam ou não

```python
#trasnformar label em letras minusculas
dados["label"] = dados["label"].str.lower()
#tirar espaços em branco no início e no final
dados["label"] = dados["label"].str.strip()
```

```python
#trocar label por 0 e 1, onde 0 é não spam e 1 é spam
dados["label"] = dados["label"].replace({"ham": 0, "spam": 1})
dados.head()
```
